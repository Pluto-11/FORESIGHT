## 2. Data Quality Checks

### Check sku_master

```python
print("Shape:", sku_master.shape)

print("\nDuplicate SKUs:")
print(sku_master["SKU"].duplicated().sum())

print("\nNegative margins:")
display(
    sku_master[sku_master["Gross_Margin_Per_Unit"] < 0]
)

print("\nInvalid cost prices:")
display(
    sku_master[sku_master["Cost_Price"] <= 0]
)

print("\nInvalid selling prices:")
display(
    sku_master[sku_master["Selling_Price"] <= 0]
)
```

### Check Sales

```python
print("Shape:", sales_daily.shape)

print("\nDuplicate Date + SKU combinations:")
print(
    sales_daily.duplicated(
        subset=["Date", "SKU"]
    ).sum()
)

print("\nUnits Sold summary:")
display(sales_daily["Units_Sold"].describe())

print("\nNegative Units Sold:")
display(
    sales_daily[sales_daily["Units_Sold"] < 0]
)


print("\nPromotion distribution:")
display(sales_daily["Promotion"].value_counts())
```

### Check Calender

```python
print("Shape:", calendar.shape)

print("\nDuplicate dates:")
print(calendar["date"].duplicated().sum())


print("\nDate range:")
print(calendar["date"].min(), "to", calendar["date"].max())

print("\nWeekend distribution:")
display(calendar["is_weekend"].value_counts())

print("\nHoliday distribution:")
display(calendar["is_holiday"].value_counts())
```

### Check inventory

```python
print("Shape:", inventory.shape)

print("\nDuplicate SKU + Snapshot Date:")
print(
    inventory.duplicated(
        subset=["SKU", "Snapshot_Date"]
    ).sum()
)


print("\nNegative Current Stock:")
display(
    inventory[inventory["Current_Stock"] < 0]
)

print("\nNegative On Order:")
display(
    inventory[inventory["On_Order"] < 0]
)


print("\nLead Time summary:")
display(inventory["Lead_Time_Days"].describe())
```

```python
sales_skus = set(sales_daily["SKU"])
master_skus = set(sku_master["SKU"])

print("Sales SKUs not in SKU Master:")
print(sales_skus - master_skus)
```

```python
inventory_skus = set(inventory["SKU"])

print("Inventory SKUs not in SKU Master:")
print(inventory_skus - master_skus)
```

```python
sku_master["Launch_Date"] = pd.to_datetime(
    sku_master["Launch_Date"]
)

calendar["date"] = pd.to_datetime(
    calendar["date"]
)

inventory["Snapshot_Date"] = pd.to_datetime(
    inventory["Snapshot_Date"]
)

sales_daily["Date"] = pd.to_datetime(
    sales_daily["Date"]
)
```

```python
print("Sales:", sales_daily["Date"].min(), "→", sales_daily["Date"].max())

print("Calendar:", calendar["date"].min(), "→", calendar["date"].max())

print("Inventory:", inventory["Snapshot_Date"].min(), "→", inventory["Snapshot_Date"].max())
```

```python
sales_dates = set(sales_daily["Date"])
calendar_dates = set(calendar["date"])

missing_calendar_dates = sales_dates - calendar_dates

print("Number of sales dates missing from calendar:")
print(len(missing_calendar_dates))

print("\nMissing dates:")
print(sorted(missing_calendar_dates)[:20])
```

```python
sales_daily["Calculated_Revenue"] = (
    sales_daily["Units_Sold"] * sales_daily["Price"]
)

sales_daily["Revenue_Difference"] = (
    sales_daily["Revenue"]
    - sales_daily["Calculated_Revenue"]
)

display(
    sales_daily[
        sales_daily["Revenue_Difference"].abs() > 0.01
    ].head(20)
)
```

### Generating a compact dataset summary

```python
summary = pd.DataFrame({
    "Dataset": [
        "SKU Master",
        "Sales Daily",
        "Calendar",
        "Inventory"
    ],
    "Rows": [
        sku_master.shape[0],
        sales_daily.shape[0],
        calendar.shape[0],
        inventory.shape[0]
    ],
    "Columns": [
        sku_master.shape[1],
        sales_daily.shape[1],
        calendar.shape[1],
        inventory.shape[1]
    ],
    "Missing Values": [
        sku_master.isnull().sum().sum(),
        sales_daily.isnull().sum().sum(),
        calendar.isnull().sum().sum(),
        inventory.isnull().sum().sum()
    ]
})

display(summary)
```
